```python
from sklearn.preprocessing import StandardScaler
```

```python
y=df['revenue']
x=df.drop(columns=['revenue'])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import RobustScaler
```

```python
scal=['employees', 'employees_year', 'revenue_year', 'net_income',
       'net_income_year', 'market_cap', 'legal_units_count',
       'direct_subsidiaries_count', 'max_hierarchy_depth', 'market_cap_day',
       'market_cap_month', 'market_cap_year']

std_scaler=StandardScaler()
x_train[scal]=std_scaler.fit_transform(x_train[scal])
x_test[scal]=std_scaler.transform(x_test[scal])
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
poly = PolynomialFeatures(degree = 2)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
lr_model = LinearRegression()
lr_model.fit(x_train_poly,y_train)
y_pred = lr_model.predict(x_test_poly)
y_pred
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

comparison.head()
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(n_estimators=100,random_state=42,n_jobs=-1)
rf_model.fit(x_train, y_train)
y_pred_rf = rf_model.predict(x_test)
y_pred_rf
print("MSE:", mean_squared_error(y_test, y_pred_rf))
print("R2 Score:", r2_score(y_test, y_pred_rf))
```

```python
feature_names = poly.get_feature_names_out(x_train.columns)
coefficients = np.asarray(lr_model.coef_).ravel()
print("Feature names:", len(feature_names))
print("Coefficients:", len(coefficients))
feature_importances = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coefficients,
    "absolute_importance": np.abs(coefficients)
}).sort_values(
    by="absolute_importance",
    ascending=False
)

print(feature_importances.head(20))
```

```python
top_features = feature_importances.sort_values(
    "absolute_importance",
    ascending=False
).head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x="feature",
    y="absolute_importance",
    data=top_features
)

plt.title("Top 10 Feature Importances")
plt.xlabel("Feature")
plt.ylabel("Importance")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
